# Baseline

```python
from pathlib import Path
import re
```

```python
JA_PATH = Path('train/ja_train')
JA_DEV = Path('dev/ja_dev')
```

## Train set

### Drug

```python
!mkdir train_drug_baseline
```

```python
out_dir = Path('train_drug_baseline')
```

```python
drugs = ['薬', # drug
         '製品' # product
        ]
```

```python
for f in JA_PATH.iterdir():
    if f.suffix != '.txt':
        continue
    with f.open() as data:
        annotations = []
        fields = data.readlines()[0].split(':')
        keyid = fields[0]
        offset = len(keyid)+1
        text = ':'.join(fields[1:])
        tid = 1
        for drug in drugs:
            for m in re.finditer(drug, text):
                annotations.append((f"T{tid}", "DRUG", m.start() + offset, m.end() + offset, drug))
                tid += 1
        sorted(annotations, key=lambda a: a[2])
        with open(out_dir / (f.stem + '.ann'), 'w') as out:
            for annotation in annotations:
                out.write(f"{annotation[0]}\t{annotation[1]} {annotation[2]} {annotation[3]}\t{annotation[4]}\n")
```

### Drug + Disorder + Katakana

```python
!mkdir train_drug_disorder_katakana_baseline
```

```python
out_dir = Path('train_drug_disorder_katakana_baseline')
```

```python
katakana = "".join(chr(i)for i in range(ord(u"\u30a1"), ord(u"\u30ff") - 4))
```

```python
katakana_re = re.compile("[%s]{2,}" % re.escape(katakana))
```

```python
for f in JA_PATH.iterdir():
    if f.suffix != '.txt':
        continue
    with f.open() as data:
        annotations = []
        fields = data.readlines()[0].split(':')
        keyid = fields[0]
        offset = len(keyid)+1
        text = ':'.join(fields[1:])
        tid = 1
        for drug in drugs:
            for m in re.finditer(drug, text):
                annotations.append((f"T{tid}", "DRUG", m.start() + offset, m.end() + offset, drug))
                tid += 1
        for m in katakana_re.finditer(text):
            annotations.append((f"T{tid}", "DRUG", m.start() + offset, m.end() + offset, m[0]))
            tid += 1
        for m in re.finditer(disorder, text):
            annotations.append((f"T{tid}", "DISORDER", m.start() + offset, m.end() + offset, disorder))
            tid += 1

        sorted(annotations, key=lambda a: a[2])
        with open(out_dir / (f.stem + '.ann'), 'w') as out:
            for annotation in annotations:
                out.write(f"{annotation[0]}\t{annotation[1]} {annotation[2]} {annotation[3]}\t{annotation[4]}\n")
```

### Drug + Disorder + Katakana + Dict

```python
# http://www.edrdg.org/wiki/index.php/JMdict-EDICT_Dictionary_Project
from bs4 import BeautifulSoup
with open('JMdict_e') as inxml:
    soup = BeautifulSoup(inxml, 'xml')

jmdict = []
for entry in soup.JMdict.find_all('entry'):
    jmdict.append(entry.reb.text.strip())
```

```python
!mkdir train_drug_disorder_katakana_dict_baseline
```

```python
out_dir = Path('train_drug_disorder_katakana_dict_baseline')
```

```python
for f in JA_PATH.iterdir():
    if f.suffix != '.txt':
        continue
    with f.open() as data:
        annotations = []
        fields = data.readlines()[0].split(':')
        keyid = fields[0]
        offset = len(keyid)+1
        text = ':'.join(fields[1:])
        tid = 1
        for drug in drugs:
            for m in re.finditer(drug, text):
                annotations.append((f"T{tid}", "DRUG", m.start() + offset, m.end() + offset, drug))
                tid += 1
        for m in katakana_re.finditer(text):
            if m[0] in jmdict:
                continue
            annotations.append((f"T{tid}", "DRUG", m.start() + offset, m.end() + offset, m[0]))
            tid += 1
        for m in re.finditer(disorder, text):
            annotations.append((f"T{tid}", "DISORDER", m.start() + offset, m.end() + offset, disorder))
            tid += 1

        sorted(annotations, key=lambda a: a[2])
        with open(out_dir / (f.stem + '.ann'), 'w') as out:
            for annotation in annotations:
                out.write(f"{annotation[0]}\t{annotation[1]} {annotation[2]} {annotation[3]}\t{annotation[4]}\n")
```
