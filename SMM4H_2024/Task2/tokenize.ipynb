# Tokenizer Error

```python
from pathlib import Path
from transformers import AutoTokenizer
```

```python
model_name = "daisaku-s/medtxt_ner_roberta"
tokenizer = AutoTokenizer.from_pretrained(model_name)
JA_PATH = Path('train/ja_train')
```

## Train Set

```python
corpus = {}
for f in JA_PATH.iterdir():
    if f.suffix == '.txt':
        with f.open() as text:
            fields = text.readlines()[0].split(':')
            keyid = fields[0]
            offset = len(keyid)+1
            data = ':'.join(fields[1:])
            if f.stem in corpus:
                corpus[f.stem]['text'] = data
                corpus[f.stem]['offset'] = len(keyid) + 1
            else:
                corpus[f.stem] = {'text': data,
                                  'offset': len(keyid) + 1,
                                  'ann': []}
    elif f.suffix == '.ann':
        with f.open() as anno:
            for line in anno:
                if line.startswith("T"):
                    fields = line.strip().split("\t")
                    if f.stem in corpus:
                        corpus[f.stem]['ann'].append((fields[1], fields[2]))
                    else:
                        corpus[f.stem] = {'ann': [(fields[1], fields[2])]}
```

```python
train_count = 0
total_count = 0
for key, anndic in corpus.items():
    text = anndic['text']
    offset = anndic['offset']
    tokenized = tokenizer(text)
    tokens = tokenized.tokens()
    for label, ner in anndic['ann']:
        tag, ts, te = label.split(' ')
        ts = int(ts) - offset
        te = int(te) - offset
        start = tokenized.char_to_token(ts)
        end = tokenized.char_to_token(te - 1)
        tokenized_ner = "".join(tokens[start: end + 1]).replace('▁', '')
        total_count += 1
        if tokenized_ner != ner:
            print(key, ner, '->', tokenized_ner)
            #for t in tokens[start: end + 1]:
            #    print("--->", t)
            train_count += 1
```

```python
print('train mismatch', train_count / total_count * 100, '%')
```
