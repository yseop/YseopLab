```python
TWEET_FILE="../train_tweets.tsv"
ANNOTATION_FILE="../train_spans_norm.tsv"
```

```python
import pandas as pd
```

## TRAIN SET

```python
tweet_df =  pd.read_csv(TWEET_FILE, sep='\t',
                        header=None, names=['id', 'tweet'],
                        dtype="string")
annotation_df = pd.read_csv(ANNOTATION_FILE, sep='\t',
                            header=None,
                            names=['id', 'cat', 'start', 'end', 'text', 'ptid'],
                            dtype={'id': "string", 'cat':"string", 'text':"string"})
merged_df = annotation_df.merge(tweet_df, on='id')
```

### Annotation Errors

```python
missaligned_ids = set()
for _, row in merged_df.iterrows():
    if row.tweet[row.start: row.end].lower() != row.text.lower():
        missaligned_ids.add(row.id)
print(len(missaligned_ids))
```

### Disable Quoting

```python
tweet_df =  pd.read_csv(TWEET_FILE, sep='\t',
                        header=None, names=['id', 'tweet'],
                        dtype="string",
                        quotechar=None,
                        quoting=3)
annotation_df = pd.read_csv(ANNOTATION_FILE,
                            sep='\t',
                            header=None,
                            names=['id', 'cat', 'start', 'end', 'text', 'ptid'],
                            dtype={'id': "string", 'cat':"string", 'text':"string"})
merged_df = annotation_df.merge(tweet_df, on='id')
```

```python
missaligned_ids = set()
for _, row in merged_df.iterrows():
    if row.tweet[row.start: row.end].lower() != row.text.lower():
        missaligned_ids.add(row.id)
print(len(missaligned_ids))
```

### Prefer to work with quoting enabled

There were some annotations that were off by 1 and the rest we cleaned manually

```python
tweet_df =  pd.read_csv(TWEET_FILE,
                        sep='\t',
                        header=None,
                        names=['id', 'tweet'],
                        dtype="string")
annotation_df = pd.read_csv(ANNOTATION_FILE,
                            sep='\t',
                            header=None, names=['id', 'cat', 'start', 'end', 'text', 'ptid'],
                            dtype={'id': "string", 'cat':"string", 'text':"string"})
merged_df = annotation_df.merge(tweet_df, on='id')
```

```python
offset = 1
for _, row in merged_df.iterrows():
    if row.tweet[row.start: row.end].lower() == row.text.lower():
        pass
    elif row.tweet[row.start: row.end].lower() != row.text.lower() and row.tweet[row.start-offset: row.end-offset].lower() == row.text.lower():
        pass
    else:
        print(row.id, row.tweet[row.start: row.end], row.text)
```
